```python
import matplotlib.pyplot as plt
import pandas as pd

# Suponiendo que tus DataFrames ya están cargados
# df100tray, df100conv, df50tray, df50conv, df20tray, df20conv
dataset_names = ['dense100', 'conv100', 'dense50', 'conv50', 'dense20', 'conv20']
# Lista de todos los DataFrames
dataframes = [tray_100, conv_100, tray_50, conv_50, tray_20, conv_20]

# Columnas que quieres graficar
columns_to_plot = ['a_1', 'a_2', 'b_1', 'b_2', 'd_1', 'd_2', 'n']

# Iterar a través de cada columna
for col in columns_to_plot:
    plt.figure(figsize=(10, 6))
    for i, (df, name) in enumerate(zip(dataframes, dataset_names)):
        plt.plot(df[col], label=f"{col}  {name}")
        plt.plot(df[f"val_{col}"], linestyle='--', label=f"val_{col}  {name}")
    plt.savefig(f"{col}_plot campvec.png")
    plt.title(f"Plot for {col}")
    plt.xlabel("Epochs")
    plt.ylabel(col)
    plt.legend()
    plt.show()
```

```python
fig, axs = plt.subplots(nrows=2, ncols=4, figsize=(20, 10))

for i, col in enumerate(columns_to_plot):
    ax = axs[i // 4, i % 4]
    for df, name in zip(dataframes, dataset_names):
        ax.plot(df[col], label=f"{col} from {name}")
        ax.plot(df[f"val_{col}"], linestyle='--', label=f"val_{col} from {name}")
    ax.set_title(f"Plot for {col}")
    ax.set_xlabel("Epochs")
    ax.set_ylabel(col)
    ax.legend()

plt.tight_layout()
plt.show()
```

```python
# Encuentra los valores mínimos y máximos globales para cada columna
global_mins = {}
global_maxs = {}
for col in columns_to_plot:
    min_vals = [df[col].min() for df in dataframes]
    max_vals = [df[col].max() for df in dataframes]
    #global_mins[col] = min(min_vals)
    global_maxs[col] = max(max_vals)

# Crear subgráficos
fig, axs = plt.subplots(nrows=2, ncols=4, figsize=(25, 15))

for i, col in enumerate(columns_to_plot):
    ax = axs[i // 4, i % 4]
    for df, name in zip(dataframes, dataset_names):
        ax.plot(df[col], label=f"{col} from {name}")
        ax.plot(df[f"val_{col}"], linestyle='--', label=f"val_{col} from {name}")
    
    # Establecer los mismos límites del eje y para cada gráfico de la misma columna
    ax.set_ylim(0.1, global_maxs[col])
    
    ax.set_title(f"{col}")
    ax.set_xlabel("Epochs")
    ax.set_ylabel(col)
    ax.legend()
plt.savefig(f"{col}_plot.png")
plt.tight_layout()
plt.show()
```
